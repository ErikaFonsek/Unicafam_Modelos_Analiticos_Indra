**3.8 Tu meta de ahorro**

```python
ahorro = 0       # Dinero que tenemos ahorrado inicialmente
meta = 5000      # Cantidad de dinero que queremos ahorrar
por_mes = 800    # Cantidad de dinero que ahorramos cada mes
meses = 0        # Contador de meses transcurridos

# Mientras el ahorro sea menor que la meta, repetimos el ciclo
while ahorro < meta:

    ahorro = ahorro + por_mes  # Sumamos el ahorro de cada mes
    meses = meses + 1          # Contamos un mes más

# Mostramos los resultados cuando termina el ciclo
print("Ahorro acumulado:", ahorro)
print("Meta de ahorro:", meta)
print("Cantidad de meses:", meses)
```

**3.9 Dividir la cuenta**

```python
cuenta = 1200    # Valor total de la cuenta
propina = 10     # Porcentaje de propina
amigos = 4       # Cantidad de personas que pagan

def con_propina(cuenta, porcentaje):

    # Calculamos el valor de la propina
    valor_propina = cuenta * porcentaje / 100

    # Sumamos la cuenta más la propina
    total = cuenta + valor_propina

    # Devolvemos el resultado
    return total

def por_persona(total, personas):

    # Dividimos el total entre la cantidad de personas
    valor_individual = total / personas

    # Devolvemos el valor que debe pagar cada persona
    return valor_individual

# Llamamos a la primera función para calcular el total
total_cuenta = con_propina(cuenta, propina)

# Llamamos a la segunda función para dividir la cuenta
pago_individual = por_persona(total_cuenta, amigos)

# Imprimimos los resultados
print("Valor de la cuenta:", cuenta)
print("Porcentaje de propina:", propina, "%")
print("Total con propina:", total_cuenta)
print("Cantidad de amigos:", amigos)
print("Cada persona debe pagar:", pago_individual)
```
